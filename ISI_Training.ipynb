```python
import scipy.stats as stats
mean = df["bmi"].mean()
std = df["bmi"].std()
print(mean)
print(std)
```

```python
#x-axis
x = np.linspace(mean - 3 * std, mean + 3 * std, 1000)
#pdf
pdf = stats.norm.pdf(x,mean, std)

plt.hist(df["bmi"], bins=30, density=True, alpha=1.0, label="Histogram")
plt.plot(x, pdf, color="red", label="Normal Distribution")
plt.xlabel("bmi")
plt.ylabel("Frequency")
plt.title("Normal Distribution for bmi")
plt.legend()
plt.grid(True, alpha=0.3, linestyle="--")
plt.show()
```
